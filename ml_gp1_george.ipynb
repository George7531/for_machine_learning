###Basics

A support vector machine is a supervised learning method used for classification and regression. The important thing to remember is it projects the data with many features into higher dimension in order to classify it with a seperator if the data is orginially non-linear

### Advantages:
* SVM can capture non-linear relationships in the data through kernals.

* It often performs effectively when the number of predictors is larger than the number of observations.

* maximizing the margin can improve generalization and it is controlled by the regulirization parameter C.

our objective is to maximize the margin between two seperate predicted classes as much as possible.

objective: maximize the margin. lets call it $ M $. And it is subject to these constraints:

$ y_k (\beta_0 + \beta_1x_{k1} + \beta_2x_{k2} + \beta_3x_{k3} ... \beta_px_{kp}) \geq M \quad \forall \quad k = (1,2,3...n)  $

This equation must be greater than or equal to the margin to seperate one class from the other in a dataset with 2 features.
and sum of all beta coefficients must be equal to 1: $ \sum_{j=1}^n \beta_j^2  =1 $, this gives the perpendicular distance from the point to the hyperplane that seperates it from points belonging to the other class. This ensures that the distance from the point to the hyperplane (which is perpendicular to it) will be atleast as large as the margin(M) itself.


This equation
$ y_k (\beta_0 + \beta_1x_{k1} + \beta_2x_{k2} + \beta_3x_{k3} ... \beta_px_{kp}) \geq M \quad \forall \quad k = (1,2,3...n)  $

one important thing to note is no data point is allowed to be present inside the margin region in max margin classifier so red balls or blue balls cannot be present within that orange dashed lines. This is where support vector classifier differs because it allows for soft margins meaning the balls or datapoints can exist within that margin allowing for a slight misclassification but robust predictive results.

support vector classifier is able to achieve it thanks to this modification to our equation by adding an error term $ (1-\epsilon) $:
$ y_k (\beta_0 + \beta_1x_{k1} + \beta_2x_{k2} + \beta_3x_{k3} ... \beta_px_{kp}) \geq M (1-\epsilon_k)  $

this $\epsilon$ allows how much `misclassification` or within margin presence of balls is allowed.

again $ \epsilon $ is subject to some constraints that is it must be greater than 0 and sum of all epsilons be less than $ C $:

$ \epsilon_k \geq 0 $

and

$ \sum_{k=1}^n \epsilon_k \leq C $

This C is an hyperparameter provided by the user, we can tune it for optimal classification. one thing to be mindufl of is larger the value $ C $, more errors we allow for our algorithm.

Now, as we move to higher dimension especially when classifying things require us to some kind of higher dimensional transformations to the feature space such as transforming the features from degree 1 to degree 2, 3, ..85 and more, how do we correctly marginalize the seperation zone? where do we put the hyperplane. The answer lies with kernal trick which if we rip open and see the mechanics is just about taking dot products between feature vectors. Mechanics of the dot product if we have two vectors is as follows

````python
Image.open('svm_pic36.jpg')
````

### Equations:

$ <a,b> = \sum_i a_i b_i  $

$ a \cdot b  =  a_1 b_1 + a_2 b_2 $

it can also be done this way: multiplying the magnitudes of the vectors a and b with cosine angle between them as follows:

$ a \cdot b = |a| |b| \cos(\theta)  $

this part: $ |a| |b| \cos(\theta) $ is more important as it finds out how similar a is to b, if $\theta$ = 0(zero) then the vectors are pointing in the same direction and if $\theta$  = 90 degrees then the dot product between these vectors is zero and they are perpendicular to each other therefore are 90 degrees and if they are pointing in the opposite direction then we'd have $\theta = 180$, so all in all dot products help figure out how similar or disimilar two vectors are.

Now using this dot product idea, we can rewrite the linear support vector as:

$ f(x) = \alpha_0 + \sum_i \alpha_i <x,x_i> $

- x -  is the new point whose class we want to predict.
- x_i - is the i-th training observation.
- <x,x_i> - we take dot product between them.

suppose the $ x_2 $ is the point (2,1) and the new unseen data point whose class we want to predict turns out to be  $ x $ =  (3,2) and we take dot product between them to see how this point (3,2) is similar to the point of $ x_2 $.

$ x^T \cdot x_2 =  [3,2] \cdot [2,1] = (3 \times 2) + (2 \times 1) = 6 + 2 = 8$

similarly if $x_3 = (5,5) $, we'd do the same process and get $ <x,x_3 > =  25 $.

$ \alpha_i $ would carry different weights for different $ <x, x_i > $ combinations and we'd use that to calculate $ \sum_i \alpha_i <x,x_i> $

$ \alpha_i$ for support vectors is non-zero, $ \alpha_i  > 0 $, it is because margin's boundary (hyperplane)  that seperates points are determined only by the closest points to it not the far-away points, those closest points form the support vector so we assign $ \alpha_i  > 0 $ and the $ \alpha_i $ for the far-away points is = 0. support vectors are the datapoints that are very close to or touching the margin.

for a hard margin support vector machine, for the datapoints lying exactly on the margin, $ y_i (\beta^T x + \beta_0) = 1 $, for them $\alpha > 0 $ and for other points $\alpha = 0 $

our linear support classifier equation rewritten: $ f(x) = \alpha_0 + \sum_{i \in Z} \alpha_i <x,x_i> $

where,
Z is a set of indices of those support vector points only, and we don't consider other points that are non-supportive in determining the location of our margin.


**kernel trick:**

kernel function is just a similarity measurement which in the basic sense is just dot products between things.

for our simple linear support vector machine we can write our kernel function as: $ K(x,x_i^u) = <x,x_i > = x^T \cdot x_i $. there are many different kinds of kernels for different kinds of tasks like radial basis kernel, polynomial kernel and so on..

The kernel trick is about putting our Kernel function in the place where we used dot product in our SVM equation: $ f(x) = \alpha_0 + \sum_i \alpha_i <x,x_i> \quad \longrightarrow f(x) = \alpha_0 + \sum_i \alpha_i K(x,x_i)  $

kernel allows us to avoid needless computations in the whole big feature space by only choosing the ones that are distinct pairs of features that's one advantage. [GeekforGeeks](https://www.geeksforgeeks.org/machine-learning/major-kernel-functions-in-support-vector-machine-svm/)

## Principal Component Analysis:

### keywords:

Dimension Reduction, 2D visualization, de-correlating features, orthogonality, Eien Decomposition.

### Advantages:
1. Reduce dimensions.
2. Decorrelated pricipal components of original features.

### Disadvantages:
Loss of interpretations. Principal components aren't the features themselves. They are the overall variances of all features combined. so, when we see a principal component showing negative or positive trend, we can't correspond it to any particular feature, so the interpretation of 'with AGE people's tendency to switch to android phones decrease' can't be done with principal components.

### Basics and Background

There was a matematician named Harold Hotelling who published his paper:[Analysis of a complex of statistical variables into principal components](https://psycnet.apa.org/record/1934-00645-001) which discusses in detail about transforming a set of variables into their principal components. This is the orgin of principal component analysis(PCA) technique.

PCA helps reduce the dimensions of data by extracting the most variance across all features(variables) of the dataset that a dataset having 100 columns(features stacked column-wise) into 2-3 columns(principal components). And these pricipal componets are uncorrelated to each other meaning if we find the correlation principal component 1 (PC1) and PC2 will be 0.

Another Area where this PCA technique could be useful is with visualizations, we can't visualize a matrix with 500 dimensions even if we plot a correlation heat map it's hard to see all 500 pairwise correlation magnitudes all at once that's where reducing or rather having only 2 principal components PC1 and PC2 explain most of the variance of the data in a 2D plot would come in handy.

I'll explain in detail about how this dimensionality reduction and de-correlation the PCA is able to achieve in mathematical equation section.

### Eigen Decomposition to get PCA components

1. Mean center the dataset/dataframe/matrix such that each column(if the features are stacked column wise) will have a mean of zero and a standard deviation of 1.
2. Compute the covariance matrix of this mean centered matrix. one beautiful thing about covariance matrix is they are symmetric meaning if A is our matrix then $ A = A^T $, a matrix is equal to its transpose.
3. Then, perform the Eigen decomposition on this covariance matrix to get this form:

    \begin{equation}
    AV = \lambda V
    \tag{1}
    \end{equation}

    where, A is our matrix and V is the eigen vectors matrix and \lambda is our vector of eigen values. if A is of size mxn, V also will of size mxn and $ \lambda $ will be of size nx1.

    professor Mike X cohen in his linear algebra book asked to imagine eigen values as keys and eigen vectors as locks to be unlocked by those keys, these eigen vectors form the principal components scaled by or strenghend by eigen values.

    the above equation (1) could be rearranged to equalling zero to take our charecteristic equation:

    \begin{equation}
    (A - \lambda I) V = 0
    \tag{2}
    \end{equation}

    $I$ stands for identity matrix of size as same that of A
    from equation (2), we want to find if there exists non-zero numbers provided we set things to null space(set to zero), and the non-zero numbers will exist only if this satisfies this condition:


    \begin{equation}
    det(A - \lambda I)  = 0
    \tag{3}
    \end{equation}

    which is called charecteristic equation, and det stands for determinant. this way we will have found eigen values; $ \lambda_1, \lambda_2, \lambda_3 .... \lambda_n $

    for each lambda values (eigen values) we will solve this equation (2): $ (A - \lambda_i I) v_i = 0  $ to get corresponding eigen vector.

    since the covariance matrix is symmetric and each column in it are orthogonal (we orthogonalized it and mean-centered it, orthogonalization btw is projecting each column in our matrix to be perpendicular to the other columns in it meaning dot product between them will be zero), inverse of eigen vectors are just their transpose, $ V^{-1} = V^T $

    our equation (1)  $ AV = \lambda V \longrightarrow A = V \lambda V^{-1} \longrightarrow A = V \lambda V^{T}$

4. After we get these eigen values and eigen vectors, to get the PCA components we need to sort the eigen values in descending order such that $ \lambda_1 \geq \lambda_2 \geq \lambda_3 \geq ....
\lambda_n $, the eigenvector associated with the largest eigenvalue points in the direction of maximum variance and it is our first principal component. Eigen vectors are the principal axes or which determines the direction principal component points at. $ K = [v_1, v_2  ..... v_n ] $

5. Finally, we have to project the data, to reduce its dimension or explain with as less columns or principal components as possible.

    $ P = X_c K $

    where,
    
    $X_c$ is our centered covariance matrix,

    $K$ is our eigne vectors,

    $P$ is our matrix with many principal components, each column in it is a principal component.

## Neural networks:

### Advantages:

**Signal perpetuation:** it helps caputre signals that have origins at a significantly long time in the past, it carries the past memories into the present.

**Unique gated mechanism:(forget gate, input gate and output gate)** Unlike standard neural networks LSTM navigates through the challenges of vanishing or explosive gradient problems effectively.

**Non-linear and interlinked signals:** LSTM effectively captures non-linear and interlinking multi variable cross-independcies that standard linear frameworks like ARIMA and ARMA often miss.

**No manual signal tracing:** LSTM automatically finds the internal latent representations of original signals such as price behaviour of an instrument without manually having to program or use an existing technical indicator which often might not fit the data as they are incoming in different points of time.

### Equations:
An LSTM cell controls data flow at time step $t$ using three distinct mathematical gates (input, forget, and output) mapped by the following continuous structural transformations:

**Forget Gate:**
$$f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + b_f)$$

This gate how much of the past memory to forget and how much to keep. $ h_{t-1} $ is the result of our previous hidden state at time t-1 and $ x_t $ is the current input fromm our feature matrix and the sigmoid function makes the result of forget gate to range between 0 and 1. An output of $0$ means "completely wipe this out of memory" while an output of $1$ means "keep this completely".

**Input Gate:**
$$i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + b_i)$$



**Candidate Cell State Update:**
$$\tilde{C}_t = \tanh(W_c \cdot [h_{t-1}, x_t] + b_c)$$

It uses the hyperbolic tangent function $\tanh$, which forces values between $-1$ and $1$.

**Updated Cell State:**
$$C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t$$

This is the the often called the cell state.$f_t \odot C_{t-1}$ point-by-point shrinks or keeps old memory ($C_{t-1}$).

$i_t \odot \tilde{C}_t$ allows the new input information.The $+$ sign blends them together to forge the new updated memory bank $C_t$.

The beauty of this architecture is the additive interaction in the updated cell state formula ($+$), because the old memory and new memory are combined linearly via addition rather than nested matrix multiplication, gradients can flow backwards through time ($T$) during backpropagation along the $C_t$ completely unhindered. This is exactly what eliminates the vanishing gradient problem found in classic RNNs.

**Output Gate:**
$$o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + b_o)$$
This gate decides what part of our memory bank ($C_t$) should be exposed to the outside world right now.

**Final Hidden Output State:**
$$h_t = o_t \odot \tanh(C_t)$$

The internal memory $C_t$ is squashed through a $\tanh$ function to stabilize its scale, and then multiplied element-wise by the output gate $o_t$.

Note: $\sigma$ represents the logistic sigmoid function, $\odot$ represents the element-wise (Hadamard) vector multiplication, $W$ represents weight matrices, and $b$ represents bias parameters.



### Model Operation Illustration

Information flows through the network pipeline in the following panache:

$$\begin{array}{rcccl}
\mathbf{\left[ \text{Input Vector } X_t \right]} & \longrightarrow & \Big( \text{Forget Gate } / \text{ Input Gate Filter} \Big) & \longrightarrow & \mathbf{\left[ \text{Cell State } C_t \text{ Update} \right]} \\
& & \Large\downarrow & & \quad\Downarrow \\
\mathbf{\left[ \text{Previous Hidden } h_{t-1} \right]} & \text{-------+} \longrightarrow & \Big( \text{Output Gate Activation} \Big) & \longrightarrow & \mathbf{\left[ \text{Next Hidden State } h_t \right]}
\end{array}$$
